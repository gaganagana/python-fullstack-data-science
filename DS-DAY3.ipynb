```python
from sklearn.linear_model import LinearRegression

x = [[35],[60],[75]] #input
y = [0,1,1] #output (0=fail, 1=pass)

model = LinearRegression()
model.fit(x,y)

print(model.predict([[50]]))
print(model.predict(([[89]])))
```
