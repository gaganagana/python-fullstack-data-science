```python
 #step 2: Import the  dataset
churn_data = pd.read_csv('Telco_Customer_Churn.csv')
dataset = pd.DataFrame(churn_data)
dataset.head()
dataset.info()
x = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

DATA CLEANING

```python
#Convert total charges to numeric values
# Defensive check: if dataset is None, re-initialize it.
# This handles cases where previous cells defining dataset were not run or kernel state was reset.
if dataset is None and 'churn_data' in locals():
    dataset = pd.DataFrame(churn_data)
    print("Warning: 'dataset' was None. Re-initialized from 'churn_data'. Please ensure cell -JQt27GhYThh is executed for proper setup.")
elif dataset is None:
    print("Error: 'dataset' is None and 'churn_data' is not defined. Please run all preceding data loading cells.")
    raise NameError("'dataset' is not defined and cannot be re-initialized.")

dataset['TotalCharges'] = pd.to_numeric(dataset['TotalCharges'], errors='coerce')
#Fill all the NaN values to median value
dataset['TotalCharges'] = dataset['TotalCharges'].fillna(dataset['TotalCharges'].median())
#Drop the unnecessary data
if 'customerID' in dataset.columns:
  dataset = dataset.drop(['customerID'], axis=1) # Removed 'inplace=True' to avoid 'None' assignment
```

#Step 5: ENCODING THE NON-NUMERICAL DATA

```python
# Defensive check: if dataset is None, re-initialize it from churn_data
if dataset is None and 'churn_data' in locals():
    dataset = pd.DataFrame(churn_data.copy()) # Use a copy to avoid modifying original churn_data if it's used elsewhere
    # Re-apply previous cleaning steps if dataset was re-initialized in this cell
    if 'customerID' in dataset.columns:
        dataset = dataset.drop(['customerID'], axis=1)
    dataset['TotalCharges'] = pd.to_numeric(dataset['TotalCharges'], errors='coerce')
    dataset['TotalCharges'] = dataset['TotalCharges'].fillna(dataset['TotalCharges'].median())
    print("Warning: 'dataset' was None upon entering kJ6O9OWz_gym. Re-initialized from 'churn_data' and cleaning steps re-applied.")
elif dataset is None:
    print("Error: 'dataset' is None and 'churn_data' is not defined. Please run all preceding data loading cells.")
    raise NameError("'dataset' is not defined and cannot be re-initialized.")

#Convert th churn data into 0 or 1
dataset['Churn'] = dataset['Churn'].map({'Yes': 1, 'No': 0})
#Convert into binary encoding
dataset = pd.get_dummies(dataset,drop_first=True)

dataset.head()
```

Step 6: TRAIN TEST SPLIT

```python
#Split the data into dependant and independant values
X = dataset.drop('Churn', axis=1)
Y = dataset['Churn']
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2,random_state=48)

```

FEATURE SCALING

```python
from sklearn.preprocessing import StandardScaler,LabelEncoder
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# STEP 8 : LOGISTIC REGRESSION
# Initialize model
log_model = LogisticRegression()
# Train model
log_model.fit(X_train, Y_train)
# Predictions
Y_pred_log = log_model.predict(X_test)
```

```python
# STEP 9 : MODEL EVALUATION

from sklearn.metrics import confusion_matrix, classification_report, roc_auc_score
# Confusion Matrix
cm = confusion_matrix(Y_test, Y_pred_log)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix")
plt.show()

# Classification Report
print(classification_report(Y_test, Y_pred_log))

# ROC-AUC Score
roc_score = roc_auc_score(Y_test, log_model.predict_proba(X_test)[:,1])
print("ROC-AUC Score:", roc_score)
```

```python
# STEP 10 : RANDOM FOREST

rf_model = RandomForestClassifier()
rf_model.fit(X_train, Y_train)
Y_pred_rf = rf_model.predict(X_test)
print(classification_report(Y_test, Y_pred_rf))
```

```python
# STEP 11 : FEATURE IMPORTANCE

# Get feature importance
importance = pd.Series(rf_model.feature_importances_, index=X.columns)
# Plot top 10 important features
importance.nlargest(10).plot(kind='barh')
plt.title("Top 10 Important Features")
plt.show()
```
