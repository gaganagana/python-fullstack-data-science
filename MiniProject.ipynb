```python
#Step 1:Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#set style for plots
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10,6)

#Step 2:Load the dataset
df = pd.read_csv("/content/Supermart Grocery Sales - Retail Analytics Dataset.csv")
df.head()

```

```python
#Check shape, columns, and data types
print("Dataset Shape:", df.shape)
print("\nColumn Names:", df.columns)
print("\nData Type:\n", df.dtypes)
#df.dtype showing the name of the particular column name

#Check for missing values
print("\nMissing Values:\n", df.isnull().sum())
#if there is missing values it will consider automatically zero

#Basic Statistics
df.describe(include='all')
```

```python
#Convert data column to datetime
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
#coerce it will handle the missing values

#Handle missing or incorrect data if any
df = df.dropna(subset=['Order Date'])
#The dropna() function in pandas is used to remove
#rows or columns that contain missing values (represented as NaN or None).

#Add  new columns for analysis
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month_name()

df.head()
```

We are doing bassic analysis and profit margine (we r finding PM for all data)

```python
#Total Sales and Profit
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
print(f"Total Sales: ₹{total_sales:,.2f}")#.2f after 2 decimal point display
print(f"Total Profit: ₹{total_profit:,.2f}")#ctrl+shift+$ to get rupees symbol

#Profit Margin
df['Profit Margin'] = (df['Profit'] / df['Sales']) * 100
df.head()
```

Streamline is used to move our project

```python
#Total sales by Category
plt.figure(figsize=(8,5))
sns.barplot(x='Category', y='Sales', data=df, estimator=sum, palette='viridis')
#colorcode(palette)
plt.title('Total Sales by Category')
plt.xticks(rotation=45)
plt.show()

#Total Profit by Category
plt.figure(figsize=(8,5))
sns.barplot(x='Category', y='Profit', data=df, estimator=sum, palette='mako')
plt.title('Total Profit by Category')
plt.xticks(rotation=45)#if we want to rotate graph
plt.show()


#Average discount by Category
plt.figure(figsize=(8,5))
sns.barplot(x='Category', y='Discount', data=df, estimator=np.mean, palette='coolwarm')
plt.title('Average Discount by Category')
plt.xticks(rotation=45)
plt.show()
```

```python
#Total Sales by Region
sns.barplot(x='Region', y='Sales', data=df, estimator=sum, palette='plasma')
plt.title('Total Sales by Region')
plt.show()

#Profit vs Sales Scatter by Region(Scatterplot is a regration line)
sns.scatterplot(x='Sales', y='Profit', data=df, hue='Region', alpha=0.7)
plt.title('Profit vs Sales by Region')
plt.show()

#Top 10 Cities by sales
top_cities = df.groupby('City')['Sales'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(8,5))
top_cities.plot(kind='bar', color='green')
plt.title("Top 10 Cities by Sales")
plt.xlabel("City")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.tight_layout()#it will adjust the graph
plt.show()
```

```python
#Monthly Sales Treand
month_sales = df.groupby('Month')['Sales'].sum().reindex(['January','February','March','April','May','June','July','August','September','October','November','December'])
#reindex name the function once again
month_sales.plot(kind='line', marker='o', color='purple')
plt.title('Monthly Sales Trend')
plt.ylabel("Sales")
plt.show()

#Yearly Profit Trend
year_profilt = df.groupby('Year')['Profit'].sum()
year_profilt.plot(kind='bar', color='orange')
plt.title('Yearly Profit Trend')
plt.ylabel("Profit")
plt.show()
```

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("/content/Supermart Grocery Sales - Retail Analytics Dataset.csv")
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
df['Year']= df['Order Date'].dt.year #automatically convert datetime
df['Month'] = df['Order Date'].dt.month_name()
```

```python
fig = px.bar(
    df.groupby('Category', as_index=False)['Sales'].sum(),
    x='Category', y='Sales',
    color = 'Category',
    title='🛒Total Sales by Category(Interactive)'
)
fig.update_layout(xaxis_title='Category', yaxis_title='Sales')
fig.show()
```

```python
fig = px.bar(
    df.groupby('Region', as_index=False)['Profit'].sum(),
    x='Region', y='Profit',
    color = 'Region',
    title='🛒Total Profit by Region(Interactive)'
)
fig.update_traces(texttemplate='%{y:.2f}', textposition='outside')
fig.show()
```

```python
fig = px.scatter(
    df, x='Sales',y ='Profit',
    color='Category',
    hover_data=['Customer Name','City','Discount'],
    title='🛒Profit vs Sales by Region(Interactive Scatter by Category)'
)
fig.show()
```

```python
fig = px.pie(
    df, values='Sales', names='Region',
    title='🛒Sales Distribution by Region',
    color_discrete_sequence=px.colors.qualitative.Set2
)
fig.show()
```

```python
#Create subplot grid: 2 rows, 3 columns
from plotly.subplots import make_subplots
import plotly.graph_objects as go

fig = make_subplots(
    rows=2, cols=3,
    subplot_titles=(
        "🛍️Sales by Category",
        "💰Profit by Region",
        "📈Monthly Sales Trend",
        "🌍Sales by Region(Pie)",
        "💵Profit vs Sales Scatter",
        "📊Average Discounr by Category"
    ),
    specs=[
        [{"type": "bar"}, {"type": "bar"}, {"type": "scatter"}],
        [{"type": "pie"}, {"type": "scatter"}, {"type": "bar"}]
    ]
)
#-------1️⃣sales by Category---------
cat_sales = df.groupby('Category')['Sales'].sum().reset_index()
fig.add_trace(
    go.Bar(x=cat_sales['Category'], y=cat_sales['Sales'],marker_color='teal',name='Sales by Category'),
    row=1, col=1
)

#--------2️⃣Profit by Region----------
reg_profit = df.groupby('Region')['Profit'].sum().reset_index()
fig.add_trace(
    go.Bar(x=reg_profit['Region'], y=reg_profit['Profit'],marker_color='orange',name='Profit by Region'),
    row=1, col=2
)

#--------3️⃣Monthly Sales Trend--------
month_sales = df.groupby('Month')['Sales'].sum().reset_index().sort_values('Month')
fig.add_trace(
    go.Scatter(x=month_sales['Month'], y=month_sales['Sales'],mode='lines+markers',line = dict(color='purple'),name='Monthly Sales Trend'),
    row=1, col=3
)
#--------4️⃣Sales by Region (Pie chart)----
fig.add_trace(
    go.Pie(labels=reg_profit['Region'],values=reg_profit['Profit'],hole=0.3,name='Sales Share'),
    row=2, col=1
)

#--------5️⃣Profit vs Sales Scatter---------
fig.add_trace(
    go.Scatter(x=df['Sales'], y=df['Profit'],
               mode='markers',
               marker=dict(size=10,color=df['Discount'],colorscale='viridis', showscale=True),
               name='Profit vs Sales '
    ),
    row=2, col=2
)

#--------6️⃣Average Discount by Category--------
cat_discount = df.groupby('Category')['Discount'].mean().reset_index()
fig.add_trace(
    go.Bar(x=cat_discount['Category'], y=cat_discount['Discount'],marker_color='green',name='Average Discount'),
    row=2 , col=3
)

#-----Layout Adjustment----------
fig.update_layout(
    title="📊Supermart Retail Sales Dahboard(Interactive)",
    showlegend=False,
    height=900, width=1400,
    title_font=dict(size=20),
    template="plotly_white"
)

fig.update_xaxes(title_text="Category", row=1, col=1)
fig.update_yaxes(title_text="Sales(₹)", row=1, col=1)
fig.update_xaxes(title_text="Region", row=1, col=2)
fig.update_yaxes(title_text="Profit (₹)", row=1, col=2)
fig.update_xaxes(title_text="Month", row=1, col=3)
fig.update_yaxes(title_text="Sales (₹)", row=1, col=3)
fig.update_xaxes(title_text="Sales (₹)", row=2, col=2)
fig.update_yaxes(title_text="Profit (₹)", row=2, col=2)
fig.update_xaxes(title_text="Category", row=2, col=3)
fig.update_yaxes(title_text="Discount (%)", row=2, col=3)

fig.show()
```
